## 1. Zeichen und Positionen
Ein String ist eine Folge von Zeichen – fast wie eine Liste. Positionen starten bei `0`, negative zählen von hinten.

```python
wort = "Python"
print(wort[0])
print(wort[-1])
print(len(wort))
```

## 2. Slicing: `text[start:ende]`
`ende` ist **nicht** mehr dabei. Fehlt `start` oder `ende`, geht es bis zum Rand.

```python
wort = "Programmieren"
print(wort[0:7])
print(wort[:3])
print(wort[7:])
print(wort[::-1])   # rückwärts!
```

## 3. Nützliche Methoden

```python
satz = "  Hallo Welt, hallo Python  "
print(satz.strip())
print(satz.lower())
print(satz.upper())
print(satz.replace("hallo", "tschüss"))
print(satz.split())
print("Welt" in satz)
print(satz.count("allo"))
```

`split()` ist besonders praktisch: Es macht aus einem Text eine **Liste**. Mit `" ".join(liste)` geht es zurück.

```python
eingabe = "Schrauben;Dübel;Nägel"
teile = eingabe.split(";")
print(teile)
print(" + ".join(teile))
```

## 4. f-Strings
Mit `f"..."` setzt du Variablen direkt in Text ein. Nach `:` kannst du formatieren.

```python
name = "Mia"
preis = 3.5
print(f"{name} zahlt {preis:.2f} €")
print(f"|{name:>10}|")   # rechtsbündig, 10 breit
print(f"{0.256:.1%}")
```
