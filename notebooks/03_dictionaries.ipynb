## 1. Wozu?
Mit zwei Listen muss man Positionen synchron halten:

```python
namen = ["Ali", "Bea", "Cem"]
punkte = [12, 17, 9]
print(namen[1], "hat", punkte[1])
```

Ein **Dictionary** speichert Paare aus **Schlüssel → Wert**. Man sucht über den Namen, nicht über die Position.

```python
punkte = {"Ali": 12, "Bea": 17, "Cem": 9}
print(punkte["Bea"])
```

## 2. Ändern, hinzufügen, löschen, prüfen

```python
punkte["Cem"] = 11        # ändern
punkte["Dana"] = 14       # neu
del punkte["Ali"]         # löschen
print(punkte)
print("Ali" in punkte)
print(punkte.get("Ali", 0))   # kein KeyError, sondern Standardwert
```

## 3. Durchlaufen

```python
for name, wert in punkte.items():
    print(f"{name}: {wert}")

print(list(punkte.keys()))
print(sum(punkte.values()))
```

## 4. Zählen mit Dictionaries – ein Klassiker

```python
text = "banane"
haeufigkeit = {}
for b in text:
    haeufigkeit[b] = haeufigkeit.get(b, 0) + 1
print(haeufigkeit)
```
