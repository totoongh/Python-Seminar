## 1. Was ist ein Modul?
Eine Python-Datei mit fertigen Funktionen. Python bringt viele mit (die *Standardbibliothek*). Mit `import` holst du sie in dein Programm.

```python
import random

print(random.randint(1, 6))
print(random.choice(["Stein", "Schere", "Papier"]))
karten = [1, 2, 3, 4, 5]
random.shuffle(karten)
print(karten)
```

## 2. Verschiedene Import-Arten

```python
import math
print(math.sqrt(16))

from math import pi
print(pi)

import datetime as dt
print(dt.date.today())
```

`from modul import name` spart Tipparbeit, aber man sieht nicht mehr, woher `name` kommt. Im Zweifel: `import modul`.

## 3. Nützliche Module

| Modul | Wofür |
|---|---|
| `random` | Zufallszahlen |
| `math` | Wurzel, Runden, π |
| `datetime` | Datum und Uhrzeit |
| `time` | `time.sleep(1)` = 1 s warten |
| `os` / `pathlib` | Dateien und Ordner |
| `json` / `csv` | Daten speichern |

```python
from datetime import date

ausbildungsende = date(2029, 7, 31)
noch = ausbildungsende - date.today()
print(f"Noch {noch.days} Tage bis zum Ausbildungsende.")
```

## 4. Eigene Module
Jede `.py`-Datei ist ein Modul. Die nächste Zelle legt eine Datei `werkzeuge.py` an – danach kannst du sie importieren:

```python
%%writefile werkzeuge.py
def brutto(netto):
    return round(netto * 1.19, 2)

def begruessung(name):
    return f"Moin {name}!"
```

```python
import werkzeuge
print(werkzeuge.brutto(100))
print(werkzeuge.begruessung("Bea"))
```

In VS Code funktioniert das genauso: zwei Dateien im selben Ordner, `import dateiname` (ohne `.py`).
