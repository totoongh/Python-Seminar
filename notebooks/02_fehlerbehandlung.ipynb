## 1. Das Problem
Führe die Zelle aus und gib `abc` ein. Was passiert?

```python
zahl = int(input("Zahl: "))
print(zahl * 2)
```

Python stoppt mit einem `ValueError`. In einem echten Programm ist das ärgerlich – der Nutzer hat sich nur vertippt.

## 2. Die Lösung: `try` / `except`
Python *versucht* den Block unter `try`. Tritt dort der genannte Fehler auf, springt es in `except`, statt abzustürzen.

```python
try:
    zahl = int(input("Zahl: "))
    print(zahl * 2)
except ValueError:
    print("Das war keine Zahl!")
```

## 3. Wiederholen, bis es passt
Kombiniert mit `while` – das Muster wirst du oft brauchen:

```python
while True:
    try:
        alter = int(input("Alter: "))
        break
    except ValueError:
        print("Bitte nur Ziffern eingeben.")
print(f"Du bist {alter} Jahre alt.")
```

## 4. Verschiedene Fehler unterscheiden

```python
def teile(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Durch 0 teilen geht nicht.")
        return None

print(teile(10, 2))
print(teile(10, 0))
```

Häufige Fehlertypen:

| Fehler | Wann? |
|---|---|
| `ValueError` | `int("abc")` |
| `ZeroDivisionError` | `5 / 0` |
| `IndexError` | `liste[99]` |
| `KeyError` | Schlüssel fehlt im Dictionary |
| `TypeError` | `"3" + 4` |

⚠️ **Niemals** nur `except:` ohne Fehlertyp schreiben – dann verschluckst du auch Fehler, die du sehen solltest.
